```python
df_clean = df.copy()
```

```python
df_clean['Item'] = df_clean['Item'].fillna('Unknown')
```

```python
mask = df_clean[['Price Per Unit', 'Quantity', 'Total Spent']].notna().all(axis=1)

(df_clean.loc[mask, 'Total Spent'] -
 df_clean.loc[mask, 'Price Per Unit'] * df_clean.loc[mask, 'Quantity']).abs().gt(0.01).sum()
```

```python
mask = df_clean['Price Per Unit'].isnull()

df_clean.loc[mask, 'Price Per Unit'] = (
    df_clean.loc[mask, 'Total Spent'] /
    df_clean.loc[mask, 'Quantity']
)
```
